````python
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-6"
````

[user guide](https://platform.claude.com/docs/en/get-started)

### Args:
**max_tokens**: The maximum number of tokens to generate before stopping.

Note that our models may stop _before_ reaching this maximum. This parameter
only specifies the absolute maximum number of tokens to generate.

Set to `0` to populate the
[prompt cache](https://platform.claude.com/docs/en/build-with-claude/prompt-caching#pre-warming-the-cache)
without generating a response.

Different models have different maximum values for this parameter. See
[models](https://platform.claude.com/docs/en/about-claude/models/overview) for
details.
**messages**:
[input examples](https://platform.claude.com/docs/en/build-with-claude/working-with-messages).

Note that if you want to include a
[system prompt](https://platform.claude.com/docs/en/build-with-claude/prompt-engineering/claude-prompting-best-practices#give-claude-a-role),
you can use the top-level `system` parameter — there is no `"system"` role for
input messages in the Messages API.

````python
# ctrl + left-click for more info on the message.create() function
response = client.messages.create(
    model=model, 
    max_tokens=1000,
    messages=[
        {
            "role": "user", 
            "content": "What are the main python libraries for data science? Be succinct."}
    ]
)
print(response)
print(response.content[0])
````

````python
# print all of response's variables 

for var in vars(response):
    value = getattr(response, var)
    print(var, "==>", value)
    if value is None:
        continue
    elif type(value) == list:
        for item in value:
            print("  -->", item)
    elif type(value) == dict:
        for key, val in value.items():
            print("  -->", key, ":", val)
    elif type(value) not in (list, dict, int, float, str, bool):
        for nested_var in vars(value):
            print("  -->", nested_var, getattr(value, nested_var))
````
